## SECTION A – THEORY


### Q1. CTE
A Common Table Expression (CTE) is a temporary result set used within a query to improve readability and modularity.

### Q2. Updatable vs Read-only Views
Views are updatable when based on a single table without aggregations. Views with joins/aggregations are read-only.

### Q3. Stored Procedures
- Reusability
- Performance optimization
- Security (controlled execution)

### Q4. Triggers
Triggers automatically execute on events like INSERT/DELETE. Example: audit logs.

### Q5. Data Modelling & Normalization
Ensures reduced redundancy, better integrity, and efficient database design.


## Create Tables

```

CREATE TABLE Products (
    ProductID INT PRIMARY KEY,
    ProductName VARCHAR(100),
    Category VARCHAR(50),
    Price DECIMAL(10,2)
);

CREATE TABLE Sales (
    SaleID INT PRIMARY KEY,
    ProductID INT,
    Quantity INT,
    SaleDate DATE,
    FOREIGN KEY (ProductID) REFERENCES Products(ProductID)
);
```

## Q6 CTE Revenue

```
WITH revenue_cte AS (
SELECT p.ProductID, p.ProductName, SUM(p.Price * s.Quantity) AS revenue
FROM Products p
JOIN Sales s ON p.ProductID = s.ProductID
GROUP BY p.ProductID, p.ProductName)
SELECT * FROM revenue_cte WHERE revenue > 3000;
```

## Q7 View

```
CREATE VIEW vw_CategorySummary AS
SELECT Category, COUNT(*) AS TotalProducts, AVG(Price) AS AveragePrice
FROM Products GROUP BY Category;
```

## Q8 Updatable View

```
CREATE VIEW vw_Product AS
SELECT ProductID, ProductName, Price FROM Products;

UPDATE vw_Product SET Price = 1300 WHERE ProductID = 1;
```

## Q9 Stored Procedure

```
DELIMITER //
CREATE PROCEDURE GetProductsByCategory(IN cat VARCHAR(50))
BEGIN
SELECT * FROM Products WHERE Category = cat;
END //
DELIMITER ;
```

## Q10 Trigger

```
CREATE TABLE ProductArchive (
ProductID INT,
ProductName VARCHAR(100),
Category VARCHAR(50),
Price DECIMAL(10,2),
DeletedAt TIMESTAMP DEFAULT CURRENT_TIMESTAMP);

DELIMITER //
CREATE TRIGGER trg_after_delete
AFTER DELETE ON Products
FOR EACH ROW
BEGIN
INSERT INTO ProductArchive(ProductID, ProductName, Category, Price)
VALUES (OLD.ProductID, OLD.ProductName, OLD.Category, OLD.Price);
END //
DELIMITER ;
```
